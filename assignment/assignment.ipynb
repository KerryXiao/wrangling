**Q3.** Open the "tidy_data.pdf" document in the repo, which is a paper called Tidy Data by Hadley Wickham.

  1. Read the abstract. What is this paper about? The paper is about data tidying which is a important part of data cleaning. The author argues that most effort spent in data analysis is cleaning the data. However, there isn't really much guidance on how to structure data effectively. The author introduces the idea of tidy data which is where each variable is a column, each observation is a row, and each type of observational unit forms a table. This structure makes it so the data is more easily manipulated, visualized, and modeled. This enables tools to work together more efficiently.
  2. Read the introduction. What is the "tidy data standard" intended to accomplish? The tidy data standard is intended to provide a standardized framework for organizing datasets so that analysis is less error prone and easier. By having a standardized method for data structure, analysts would not have to worry about data cleaning solutions for each individual dataset. The standard also makes it so it is easier to develop analysis tools that work together which would reduce the need to constantly reformat data between steps of an analysis.
  3. Read the intro to section 2. What does this sentence mean: "Like families, tidy datasets are all alike but every messy dataset is messy in its own way." What does this sentence mean: "For a given dataset, it’s usually easy to figure out what are observations and what are variables, but it is surprisingly difficult to precisely define variables and observations in general." The first sentence means tidy datasets share a common structure, while messy datasets are disorganized in many different ways. Since tidy datasets all follow the same rules, they are predictable. However, since messy datasets do not, they are unpredictable in the ways they are disorganized. The second sentence means that people can intuitively identify what are observations and variables in a dataset, but there is not universal rule that would work in all contexts. What counts as a variable versus a value depends on how the data will be analyzed and what questions are being asked.
  4. Read Section 2.2. How does Wickham define values, variables, and observations? Values are individual data points such as numbers or strings. Variables are collections of values that measure the same attribute across units such as height and temperature. Observations are collections of values measured on the same unit such as one person, one day, one event, etc.
  5. How is "Tidy Data" defined in section 2.3? "Tidy Data' is defined as a dataset where each variable forms a column, each observations forms a row, and each type of observational unit forms a table. Any dataset that violates at least on of the rules above are messy.
  6. Read the intro to Section 3 and Section 3.1. What are the 5 most common problems with messy datasets? Why are the data in Table 4 messy? What is "melting" a dataset? The five most common problems with messy datasets are that column headers are values, not variable names; multiple variables are stored in a single column; variables are stored in both rows and columns; multiple types of observational units are stored in one table; and one observational unit is stored across multiple tables. Table 4 is messy because income levels are stored as column headers when they are actually values of the income variable. This makes it more difficult to analyze the income variable.
  7. Why, specifically, is table 11 messy but table 12 tidy and "molten"? Table 11 is messy because variables are stored in both columns and rows, column names represent values, and the element column stores variable names instead of data. Table 12 is "molten" because the wide day columns have been stacked into rows. Table 12 is "tidy" because each variable has its own column and each row represents one day's observation.
  8. Read Section 6. What is the "chicken-and-egg" problem with focusing on tidy data? What does Wickham hope happens in the future with further work on the subject of data wrangling? The "chicken-and-egg" problem is that tidy data are only useful if tidy tools exist, and tidy tools are only practical if data are already tidy. This interdependence makes it hard too improve either data structures or tools independently. Wickham hopes that future work will develop better theories of data structure, create improved tools for data wrangling, incorporation insights from cognitive science and human computer interaction, and expand beyond tidy data to alternative representations when appropriate. He emphasized that tidy data is not the final solution, but a foundation for future progress in data analysis workflows.
